```python
# Aaron Vincent
# File Name: GPAprogram.py
# Description: Enter a student name and GPA to see if they made the dean's list 
#              or honor roll
        

def main():
    print("Welcome to the Student Qualification App")
    print("----------------------------------------")
    
    while True:
        
        last_name = input("\nEnter the student's last name (or 'ZZZ' to quit): ")
        
        
        if last_name.upper() == 'ZZZ':
            print("Exiting application. Goodbye!")
            break
            
        
        first_name = input("Enter the student's first name: ")
        
        
        try:
            gpa = float(input("Enter the student's GPA: "))
        except ValueError:
            print("Invalid input. Please enter a numerical GPA.")
            continue
            
       
        if gpa >= 3.5:
            print(f"Congratulations! {first_name} {last_name} has made the Dean's List.")
        elif gpa >= 3.25:
            print(f"Congratulations! {first_name} {last_name} has made the Honor Roll.")
        else:
            print(f"{first_name} {last_name} did not make the Dean's List or Honor Roll this time.")

if __name__ == "__main__":
    main()

```
